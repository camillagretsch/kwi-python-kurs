# Projekt Turtlerennen

Kennen Sie das Ravensburger-Spiel «Tempo, kleine Schnecke?» Es gibt zahlreiche Varianten dieses Spiels. 

Sie sollen ein solches Spiel programmieren. Wie das Spiel aussehen soll, sehen Sie in diesem [Video](https://youtu.be/cBHI7rNMtnc).

## Vorgaben

1. **Spielfeld**: Das Spielfeld wird von der Turtle gezeichnet und soll aus senkrechten Linien oder Feldern bestehen.
2. **Schildkröten**:
    * Die Schildkröte, die das Spielfeld gezeichnet hat, soll als Schiedsrichterin bei der Ziellinie sitzen und Schiedsrichterin sein. 
    * verschiedenfarbige Schildkröten werden erstellt und an den Anfang des Spielfelds (Startposition) gesetzt.
3. **Wetten** Über die Benutzereingabe soll auf eine Schildkröte getippt werden. (Alternativ kann zum Tippen auch eine Schildkröte angeklickt werden).
4. **Rennen**: Per Runde geht jede Schildkröte um eine zufällige Anzahl Schritte vorwärts.  
   Simulieren Sie einen Würfel, indem Sie Zufallszahlen im Bereich von 1 bis und mit 6 generieren.
5. **Ende des Rennens**: Das Rennen endet, wenn eine Schildkröte die Ziellinie überschritten hat.
6. **Ausgabe**: Eine Nachricht wird ausgegeben, ob die Schildkröte, auf die getippt wurde, gewonnen hat.

## Vorgehen

Wie Sie vorgehen, ist Ihre Entscheidung. Falls Sie drauf los programmieren möchten, können Sie dies gerne tun.

Falls Sie aber gerne ein paar Anhaltspunkte möchte, können Sie dieser Anleitung folgen.

1. **Spielfeld –** Zeichnen Sie mit der Turtle Ihr Spielfeld.  
   Hinter der Zeichenfläche verbirgt sich ein Koordinatensystem, der Ursprung ist in der Mitte. Die Koordinaten reichen somit von (-250,-125) bis (250, 125)
   
   <img src="./bilder/06_koordinatensystem-tScreen.png" style="height:250px"/>
   
2. **Turtle klonen –** Die erste Turtle erstellt nur das Spielfeld und spielt nicht mit  
   Mit Hilfe des Befehls `clone()` können Sie die Turtle klonen.  
   Sie können der neuen Turtle einen Namen geben und ihre Eigenschaften bearbeiten, z.B.:
```Python
rot = t.clone()  # Turtle klonen, die neue Turtle «heisst» rot  
rot.color("red") # Farbe der Turtle rot ändern
rot.name = "Rot" # Für die Turtle rot die Eigenschaft name auf den Wert "Rot" setzen
```
   **Tipp**: *Klonen Sie erst einmal **eine** Turtle und programmieren Sie das ganze Spiel für eine Turtle.*
   
3. **Mit der geklonten Turtle arbeiten**  
   Bisher hatten Sie *eine* Turtle `t`, auf die Sie Befehle angewandt haben. Nun haben Sie *mehrere* Turtles (im Beispiel `t` und `rot`).
    Wenn Sie nun wollen, dass sich die Turtle `rot` fortbewegt, müssen Sie `forward()` nicht auf `t` anwenden, sondern auf `rot`:
    ```
    rot.forward(50)
    ```
4. **Mit Koordinaten arbeiten und die Turtle «platzieren»**  
   Da dem Spielfeld ein Koordinatensystem zugrundeliegt, weiss die Turtle immer, wo sie sich befindet. Auf diese Information können Sie mit Hilfe der Turtle-Funktionen `xcor()` und `ycor()` zugreifen.  
   Ausserdem können Sie die Turtle mit Hilfe der Funktion `goto(x,y)` an den gewünschten Punkt mit den Koordinaten `(x,y)` setzen:
   
```Python
    rot.goto(25,-42)   # Turtle rot an den Punkt mit der Koordinate (25,-42) setzen.
    rot_x = rot.xcor() # Der Variable rot_x wird der Wert der x-Koordinate von rot zugewiesen, also 25
    rot_y = rot.ycor() # Der Variable rot_y wird der Wert der y-Koordinate von rot zugewiesen, also -42
```
5. **Spielzug**  
   Jede Turtle, die mitspielt, soll sich jeweils um eine Würfelzahl vorwärts bewegen, dann ist die nächste Turtle dran.  
   Für den Spielzug werden Sie denselben Code für mehrere Turtles schreiben müssen. 

6. **Game-Schleife**

Von Games sind wir uns gewohnt, dass sie so lange laufen, bis wir gewonnen haben oder die Zeit um ist.

Die Spielzüge sollen also immer ständig wiederholt werden, bis eine Schildkröte die Ziellinie überschritten hat. Bei Games sorgt dafür die sogenannte «Game-Schleife», eine while-Schleife, die so lange läuft bis das Game fertig ist. Bei while-Schleifen haben Sie gelernt, dass Sie sich um den Zustand der Schleife kümmern müssen. In der Regel geschieht dies über eine Variable, z.B. `spiel_laeuft` – einem Boolean, der am Anfang des Spiels auf `True` gesetzt wird und wenn das Spiel fertig ist, auf `False`.

Die Game-Schleife sieht somit wie folgt aus:

```Python
    spiel_laeuft = True
    while spiel_laeuft:
        # Spielzug alle Turtles machen einen Spielzug
        # Hat die Turtle die Ziellinie überschritten?
            spiel_laeuft = False # Die Turtle hat gewonnen
```

Wie immer müssen Sie auch hier die Zelle ausführen, die den Befehl `prepareTurtleGraphics()` definiert.

````python
# Führen Sie diese Zelle aus, damit Ihnen die Turtle zur Verfügung steht
import ipyturtle3 as turtle

def prepareTurtleGraphics(width=500, height=250):
    myCanvas = turtle.Canvas(width=width, height=height)
    display(myCanvas)
    turtleScreen = turtle.TurtleScreen(myCanvas)
    turtleScreen.clear()
    t = turtle.Turtle(turtleScreen)
    t.shape("turtle")
    # t.speed(0) # <--- Einkommentieren, wenn Sie die Turtle schneller machen wollen...
    turtleScreen.bgcolor("floralwhite")
    return (t, turtleScreen)

print("Nun ist der Befehl prepareTurtleGraphics() definiert und Sie können Turtlegrafiken erstellen.")
````
