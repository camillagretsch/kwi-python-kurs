## 2. Auf Elemente in Listen zugreifen

Es ist nicht ganz so einfach, auf Elemente einer Liste zuzugreifen wie auf den Wert einer Variable.

Listen sind **indexiert**, das heisst die Positionen der Elemente in der Liste sind durchnummeriert. Der **Zugriff auf Elemente von Listen** erfolgt über die Indizes (Plural für Index) zwischen eckigen Klammern.

👉 Das erste Element einer Liste `liste` hat den Index **0**.

**Beispiel**: Liste `fruechte:`

<img src="./bilder/2-1_liste_grafik.png" width="50%" />

```Python
fruechte = ["Aepfel", "Bananen", "Trauben", "Orangen", "Birnen"]
fruechte[0] # 1. Element der Liste fruechte --> 'Aepfel'
fruechte[1] # 2. Element --> 'Bananen'
fruechte[2] # --> 'Trauben'
fruechte[3] # --> 'Orangen'
fruechte[4] # --> 'Birnen'
```

Zurück zur Liste `dreierreihe`:
```Python
dreierreihe[0] # 3 (erstes Element)
dreierreihe[4] # 15 (fünftes Element)
```

**⁉️Wichtig:** Die Zahl innerhalb der eckigen Klammern wird Index genannt. Mit einem Index i kann man auf die einzelnen Elemente der Liste zugreifen. fruechte[i] liefert das Element aus der Liste fruechte an der Position i. Bei einer Liste mit n-Elementen hat das letzte Element die Position n-1 oder anders gesagt: i = len(fruechte) - 1

### 2.2. Zugriff auf Bereiche einer Liste – «Teilbereichsoperator» [start:stop:step]

Es ist auch möglich, auf Bereiche einer Liste zuzugreifen. Listen lassen sich mit ':' aufteilen (splitten). Man spricht dabei auch von «Slicing».

Der **Teilbereichsoperator** verwendet Defaultwerte (das sind Standardwerte, die verwendet werden, wenn nichts anderes angegeben ist): Wird nichts angegeben, wird die *ganze* Liste mit einer Schrittlänge von `step=1` ausgegeben. `start` entspricht somit dem Index 0, `stop` der Länge. Somit entspricht `liste[:]`der ganzen Liste.

Am Beispiel der Monate:

* alle Monate:
  ```Python 
  print(monate[:]) 
  ```  
  **Ausgabe:** \['Januar', 'Februar', 'März', 'April', 'Mai', 'Juni', 'Juli', 'August', 'September', 'Oktober', 'November', 'Dezember'\]

* die zweite Jahreshälfte:  
  ```Python 
  print(monate[len(monate)//2:]) # Ganzzahlige Division (//) weil Indizes ganzzahlig sein müssen.
  ```  
  **Ausgabe:** \['Juli', 'August', 'September', 'Oktober', 'November', 'Dezember'\]
  
  Hier entspricht `start` dem Index `len(monate)//`, `stop` der Länge und `step=1`.

### 2.3. 🛑 IndexError für Zugriff auf Elemente ausserhalb einer Liste

Sie wissen nun, wie Sie auf Elemente in Listen zugreifen können. Aber was geschieht, wenn Sie auf ein Element zugreifen, das sich gar nicht in der Liste befindet?

Nehmen Sie an, Sie haben die folgende Liste `profile = ["A", "N", "M"]`, welche die an der KWI angebotenen Profile beinhaltet. Die Länge von `profile` ist 3. Da das erste Element am Index (an Position) 0 steht, wird das letzte Element an Position 2 stehen, nicht an Position 3. Dies ist ein Denkfehler, der oft gemacht wird. Dafür gibt es auch eine spezielle Fehlermeldung, den **IndexError**. Er tritt auf, wenn auf ein Element zugegriffen wird, dessen Index nicht existiert:

```Python
profile = ["A", "N", "M"]
anzahl_profile = len(profile) # anzahl_profile wird der Wert 3 zugewiesen, 
                              # da die Liste 3 Elemente enthält

profile[anzahl_profile] # --> IndexError, da das letzte Element den Index anzahl_profile-1 hat.
```

Bisher haben Sie für for-Schleifen die `range()`-Funktion verwendet, um Bereiche zu erstellen, die von der Schleifenvariable durchlaufen werden:

```Python
for schleifenvariable in range(anzahl_durchgaenge): # Schleifenkopf
    # Schleifenkörper
```

Während die Schleife ausgeführt wird, durchläuft die Schleifenvariable `schleifenvariable` den Bereich `range(anzahl_durchgaenge)` und nimmt somit die Werte von `0` bis und *ohne* `anzahl_durchgaenge` an.

## 4. Listen verändern

Neben dem Zugriff auf Listenelemente ist es auch möglich, Listen zu verändern, denn Listen sind keine starren Konstrukte.

### 4.1. Einfügen und Löschen von Listenelementen

Es ist möglich, Listenelemente einzufügen oder zu löschen.

Hierzu können Sie die folgenden Funktionen verwenden:

* `append(element)`: Hängt ein Element `element` am Ende der Liste an.
* `pop(index)`: Löscht das Element am Index `index` und gibt es zurück.
* `insert(index, element)` Fügt ein Element `element` an der Stelle `index` in die Liste ein.

#### 4.1.1. Beispiel

Läuft Ihnen auch immer die Zeit davon? 

Die Liste `woche` enthält die Tage einer Woche:

```Python
woche = ["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"]
```

Sie möchten sich nun einen Gefallen machen und die Woche um einen Tag verlängern. 

Dazu fügen Sie am Ende der Liste den «Bonustag» `Bonus` ein:
```Python
woche.append("Bonus") 

# --> woche: ["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So", "Bonus"]
```

Das Leben ist hart, der «Bonustag» muss wieder weg, weil er leider doch keinen Platz in der Woche hat 😞. Die Funktion `pop()` gibt den Wert des gelöschten Elements zurück. Sie können den gelöschten Wert also in einer Variable speichern:
```Python
geloescht = woche.pop(len(woche)-1) 

# --> woche: ["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"], 
#     geloescht: "Bonus" 
```
