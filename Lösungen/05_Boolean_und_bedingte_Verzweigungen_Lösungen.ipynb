## 1. Aussagenlogik und Wahrheitswerte

George **Boole** (1815-1864) hat als erster erkannt, dass die Aussagenlogik als eine Algebra aufgefasst werden kann, die zwei Elemente hat (heute als die zwei Wahrheitswerte bezeichnet) ([Quelle: Wikipedia](https://de.wikipedia.org/wiki/George_Boole)). Nach ihm ist die Boolesche Algebra benannt. Sie bildet die Basis für die Digitaltechnik.

### 1.1. Wahrheitswerte (Datentyp Boolean)

Zu Ehren von George Boole werden Wahrheitswerte «Booleans» genannt. Sie können nur zwei Werte annehmen: `True` (wahr) und `False` (falsch). Da es nur zwei Werte sind, sind sie in der digitalen Welt problemlos abbildbar: 

* `False` steht für Null (keine Spannung messbar, Schalter nicht gedrückt, ...) und 
* `True` steht für 1 (eine genügend grosse Spannung ist messbar, Schalter gedrückt)

Sie haben bereits gelernt, dass Transistoren sich wie Schalter verhalten:

Ist der Schalter gedrückt, brennt das Licht (💡) (Wert `1`), ist der Schalter nicht gedrückt, brennt kein Licht (die Glühbirne leuchtet nicht; Wert `0`).

Obwohl sie als Werte `0` und `1` gespeichert werden, haben Wahrheitswerte in den meisten Programmiersprachen Namen – meistens «false» und «true». In Python werden diese gross geschrieben:

```Python
False 
True
```

#### Wahrheitswerte und die Turtle

Es gibt auch Turtlefunktionen, die Wahrheitswerte zurückgeben, z.B. die Funktionen `isdown()`, um zu sehen, ob der Stift schreibt oder `isvisible()` um zu prüfen, ob die Turtle sichtbar ist.

````python
# Führen Sie diese Zelle aus, damit Sie weiter unten Turtlegrafiken machen können...

import ipyturtle3 as turtle

def prepareTurtleGraphics(width=500, height=250):
    myCanvas = turtle.Canvas(width=width, height=height)
    display(myCanvas)
    turtleScreen = turtle.TurtleScreen(myCanvas)
    turtleScreen.clear()
    t = turtle.Turtle(turtleScreen)
    t.shape("turtle")
    turtleScreen.bgcolor("floralwhite")
    return (t, turtleScreen)
````

````python
(t, tScreen) = prepareTurtleGraphics()

t.penup()
print("Zeile 4:", t.isdown())
t.pendown()
print("Zeile 6:", t.isdown())

t.hideturtle()
print("Zeile 9:", t.isvisible())
t.showturtle()
print("Zeile 11:", t.isvisible())
````

### 1.3. Vergleichsoperatoren (Relationale Operatoren)

Im Beispiel oben wurden die beiden Werte `1` und `4` mit den Operatoren `>` «grösser als» und `<` «kleiner als» verglichen. Diese Operatoren sind **Vergleichsoperatoren**, auch «relationale Operatoren» genannt. Sie dienen dazu, zwei Werte miteinander zu *vergleichen*. 

Sie kennen Vergleichsoperatoren bereits aus dem Mathematikunterricht. In Python werden sie nicht alle gleich geschrieben wie in der Mathematik.

Hier finden Sie eine Übersicht:

| Operator | Beschreibung | Ausdruck | Gibt `True` zurück, wenn...
|:---------:|:---------|:---------|:---------|
| **`==`**| gleich | `a == b` | `a` und `b` denselben Wert haben. |
| **`!=`**| ungleich | `a != b` | `a` und `b` ungleiche Werte haben. |
| **`<`** | kleiner als | `a < b`  | `a` kleiner als `b` ist. |
| **`<=`** | kleiner gleich | `a <= b`  | `a` kleiner oder gleich `b` ist. |
| **`>`** | grösser als | `a > b`  | `a` grösser als `b` ist. |
| **`>=`** | grösser gleich | `a >= b`  | `a`grösser oder gleich `b` ist. |

### 1.4. Logische Operatoren

 Genauso wie arithmetische Operatoren («Punkt vor Strich») haben auch logische Operatoren eine Rangfolge. Im Zweifelsfall oder um das Ganze übersichtlicher zu machen, können Sie Klammern verwenden, auch wenn diese nicht nötig sind.

In Python werden die logischen Operatoren wie folgt geschrieben (Rangfolge: von oben nach unten): 

| Operator  | Beschreibung                | Ausdruck  | Gibt wahr (`True`) zurück, wenn...   |
|:---------:|:----------------------------|:----------|:-------------------------------------|
| **`not`** | NOT; «nicht»                | `not a`   | `a` `False` ist.                     |
| **`and`** | AND; «and«                  | `a and b` | `a` und `b` `True` sind.             |
| **`or`**  | OR; «oder» (einschliessend) | `a or b`  | `a` oder `b` oder beide `True` sind. |


### 1.5. Wahrheitstabellen

Eine Wahrheitstabelle bildet alle Kombinationen der Eingänge und die erwartete Ausgabe einer logischen Funktion ab. Die Spalten links repräsentieren somit alle Eingänge und die Spalte rechts den Ausgang.

Damit alle Kombinationen der Eingänge enthalten sind, werden sie zu Binärzahlen kombiniert und durchgezählt. Dabei wird für `False` der Wert 0 eingetragen und für `True` der Wert 1. 

Bei zwei Eingängen sieht dies so aus:

| Eingang A | Eingang B | Ausgang |
|:---------:|:---------:|:-------------|
| 0 | 0 |||   |
| 0 | 1 |||   |
| 1 | 0 |||   |
| 1 | 1 |||   |

| Eingang | Ausgang |
|:-------:|:-------:|
| 0       | 1       |

### 2.1. Programmablauf

Indem Sie steuern, welcher Codeblocke ausgeführt wird wenn eine Bedingung erfüllt ist oder nicht, übernehmen Sie die Kontrolle über den Programmablauf und weichen von der üblichen «Abarbeitung der Befehle von oben nach unten» ab. Die sogenannte **bedingte Verzweigung** ist damit wie die Schleife eine ***Kontrollstruktur***.

***
Sie haben im letzten Notebook mit der *for-Schleife* bereits eine Kontrollstruktur kennengelernt. Schleifen erlauben es, ganze Codeblöcke automatisch zu wiederholen. ### 2.2. Syntax der bedingten Verzweigung

Eine bedingte Verzweigung wird durch das reservierte Wort **`if`** gekennzeichnet und folgt dem Schema

> *«**Wenn** Bedingung-erfüllt ist, **dann** mache, was nur gemacht werden soll, wenn die Bedingung erfüllt ist»*.

Schauen Sie sich nun den Python-Code der beiden Programme aus dem Beispiel an.

<img src="./bilder/04_verzweigungen-code.svg" />

Hier sehen Sie wiederum auf der linken Seite eine *einseitige Verzweigung* und der rechten Seite eine *zweiseitige Verzweigung*.

#### 2.2.1. Einseitige Verzweigung

```Python
if Bedingung:
    # Block, der ausgeführt wird, wenn die Bedingung erfüllt ist
```

Das ist die einfachste Form einer Verzweigung: Es wird eine Bedingung geprüft und wenn diese erfüllt ist, wird ein Block ausgeführt. Danach geht das Programm normal weiter.

Die **Bedingung**, die geprüft wird, folgt auf das reservierte Wort **`if`**. Danach folgt wie bei der for-Schleife ein **Doppelpunkt**. Er bedeutet wiederum «Was nun folgt, gehört zusammen». Der Code, der ausgeführt wird, wenn die Bedingung erfüllt ist, wird **eingerückt**. Der Code, der nach der bedingten Verzweigung ausgeführt werden soll, ist nicht mehr eingerückt.

**Beispiel**

```Python
if regen == True: # Da regen ein Boolean ist, könnten Sie auch nur 'if regen:' schreiben
    print("Es regnet.")
print("Viel Spass draussen!") # nach der Verzweigung (wird auf jeden Fall ausgeführt)
```

#### 2.2.2. Zweiseitige Verzweigung

Es ist auch möglich, Code zu schreiben, der nur ausgeführt wird, **wenn die Bedingung nicht erfüllt ist** (also nach `False` auflöst). Dies wird mit **`else:`** gekennzeichnet. Was *sonst* ausgeführt werden soll, ist eingerückt.

```Python
if Bedingung:
    # Block, der ausgeführt wird, wenn die Bedingung erfüllt ist
else:
    # Sonst-Block, der ausgeführt wird, wenn die Bedingung nicht erfüllt ist
```

**Beispiel**

```Python
regen = input("Regnet es? ")
if regen == "ja":
    print("Es regnet.")
else:
    print("Es regnet nicht.")
print("Viel Spass draussen!")
```

Es ist auch möglich, noch weitere Bedingungsprüfungen einzubauen. Das reservierte Wort dafür ist **`elif`**:

```Python
if Bedingung:
    # Block, der ausgeführt wird, wenn die Bedingung erfüllt ist
elif zweite Bedingung:
    # Block, der ausgeführt wird, wenn die zweite Bedingung erfüllt ist
# es darf mehrere elif-Blöcke geben:
elif dritte Bedingung:
    # Block, der ausgeführt wird, wenn die dritte Bedingung erfüllt ist
```

#### 2.2.4 Sauber einrücken, sonst knallt's!

Wie der Schleifenkörper bei den For-Schleifen müssen bei bedingten Verzweigungen die zusammengehörenden Codeblöcke eingerückt werden. Wichtig ist dabei, dass alles, was zusammengehört gleich eingerückt wird.

Das **Einrücken ist in Python sehr wichtig**! Der Doppelpunkt zeigt zwar an, dass ein Block anfängt, aber nicht, was alles dazugehört. Dadurch, dass in Python Zusammengehörendes durch Einrücken gekennzeichnet wird, wird der Code auch schön übersichtlich.

Wenn nicht sauber eingerückt wurde, gibt es den sogenannten **IndentationError**.

Das folgende Programm lässt eine weisse Turtle mit eingezogenem Stift auf einem schwarzen Zeichenbereich 20 mal an einen zufälligen Punkt(x,y) gehen. Bei jedem Punkt wird sie sichtbar und hinterlässt einen Abdruck. Dann wird sie wieder unsichtbar.

a) Überlegen Sie sich, wo Sie die Farbe ändern müssen, damit die Turtle einen farbigen Abdruck hinterlässt und keinen weissen.


b) Färben Sie dann die Turtle so ein, dass die Punkte im  
1. Quadranten (x > 0 und y > 0) gelb sind  
2. Quadranten (x < 0 und y > 0) rot sind  
3. Quadranten (x < 0 und y < 0) grün sind  
4. Quadranten (x > 0 und y < 0) blau sind  
und die Punkte   
auf den Achsen (x = 0 oder y = 0) weiss sind.

````python
# Verwenden Sie für Aufgabe 8 diesen Code:

import random
(t, tScreen) = prepareTurtleGraphics()

# Vorbereitungen
tScreen.bgcolor("black") # Hintergrund einfärben
t.shape("circle")        # Form der Turtle: Kreis statt Schildkröte
t.turtlesize(0.5, 0.5)   # Grösse der Turtle (halb so gross wie normal)
t.penup()                # Stift einziehen

t.color("white")         # braucht es nachher nicht unbedingt

for i in range(20):
    # Turtle verstecken
    t.hideturtle()
    
    # Turtle an zufällige Position bringen:
    x = random.randint(-240,240)
    y = random.randint(-120,120)
    t.setpos(x,y)
    
    # Abdruck machen
    t.showturtle()
    t.stamp()
````

````python
# Lösung Aufgabe 8:

import random
(t, tScreen) = prepareTurtleGraphics()

# Vorbereitungen
tScreen.bgcolor("black") # Hintergrund einfärben
t.shape("circle")         
t.turtlesize(0.5,0.5)    # Grösse der Turtle verändern
t.penup()

for i in range(20):
    t.hideturtle()
    # Turtle an zufällige Position bringen:
    x = random.randint(-240,240)
    y = random.randint(-120,120)
    t.setpos(x,y)

    # Sie müssen die Turtle einfärben, bevor sie einen Abdruck hinterlässt 
    # und am besten bevor sie sichtbar wird (Farbwechsel).
    if x >= 0 and y >= 0:
        t.color("yellow")
    elif x < 0 and y >= 0:
        t.color("red")
    elif x < 0 and y < 0:
        t.color("green")
    elif x > 0 and y < 0:
        t.color("blue")        
    else:
        t.color("white")        
    t.showturtle()
    t.stamp()
````
