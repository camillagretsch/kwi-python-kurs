# Die while-Schleife

In diesem Notebook werden wir uns mit einem neuen Schleifentyp auseinandersetzen, der **while-Schleife**. Sie ist ein Mix aus der if-Verzweigung und der for-Schleife. Am Ende dieses Notebooks sollten Sie folgende Lernziele erreicht haben:
___
### Lernziele

* Sie können einen while-Schleife programmieren
* Sie können entscheiden, ob ein Programm mit einer while-Schleife terminiert.
* Sie können abwägen, wann eine Aufgabe besser mit einer for-Schleife und wann besser mit einer while-Schleife gelöst wird.
___

Ähnlich wie die For-Schleife dient auch die while-Schleife dazu, Programmcode mehrfach auszuführen. Während bei der For-Schleife aber die Schleifenvariable angibt, wie oft die Schleife ausgeführt wird, ist es bei der while-Schleife eine **Abbruchbedingung**, welche die Schleife beendet.

Dies kann Vor- und Nachteile haben. Einerseits können Probleme gelöst werden, welche keine fixe Anzahl Schritte benötigen, sondern wegen einer anderen Bedingung terminieren (abbrechen). Anderseits kann, wenn falsch angewandt, das Programm ewig laufen. Ein Beispiel aus dem Alltag könnte das folgende sein:

Man möchte Milch auf 80 Grad erhitzen auf dem Herd. Man weiss aber nicht genau wie lange dies dauert. Man hat aber ein Thermometer, wo man die Temperatur ablesen kann. 

So könnte das Pseudocode/Python Programm dazu aussehen:
```Python

Milch auf den Herd stellen
t = Temperatur vom Thermometer
while(t < 80):
    1min warten
    t = temperatur auf Thermometer
Milch vom Herd nehmen
```

⚠️ Wichtig bei der while-Schleife ist, dass die Bedingung irgendwann erfüllt sein muss, damit die Schleife abgebrochen werden kann. 

Im obigen Beispiel steht z. B. nirgends, dass die Herdplatte eingeschaltet ist. Wenn das vergessen ginge und man den Code oben ausführen würde, würde man ewig warten. Die Schleife würde *endlos* weiterlaufen.

***

````python
import ipyturtle3 as turtle

def prepareTurtleGraphics(width=500, height=250):
    myCanvas = turtle.Canvas(width=width, height=height)
    display(myCanvas)
    turtleScreen = turtle.TurtleScreen(myCanvas)
    turtleScreen.clear()
    t = turtle.Turtle(turtleScreen)
    t.shape("turtle")
    turtleScreen.bgcolor("floralwhite")
    return (t, turtleScreen)

print("Der Befehl prepareTurtleGraphics() ist nun definiert.")
````

In der Mathematik gibt es ein berühmtes, bis heute ungelöstes Rätsel: die Collatz-Vermutung.

Sie besagt, dass jede natürliche Zahl irgendwann bei 1 endet, wenn folgende Schritte wiederholt werden:

* Falls die Zahl gerade ist, wird sie durch 2 geteilt. 
* Ist die Zahl ungerade, wird sie mit 3 multipliziert und anschliessend 1 addiert.

Computer haben diese Regeln bereits für sehr viele Zahlen überprüft. Die Vermutung stimmte dabei jedes Mal. Das ist jedoch noch kein mathematischer Beweis. Dafür müsste bewiesen werden, dass die Vermutung für jede natürliche Zahl gilt.

Bis heute gibt es auch keine allgemeine Formel, mit der sich berechnen lässt, wie viele Rechenschritte eine Zahl benötigt, bis sie bei 1 ankommt.
